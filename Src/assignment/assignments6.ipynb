```python
#Task1
def bank_account(initial_balance):
    balance = initial_balance
    def deposit(amount):
        nonlocal balance
        balance += amount
        print(f"Deposited {amount}. New balance: {balance}")
    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            print(f"Withdrew {amount}. New balance: {balance}" )
        else:
            print("Insufficient funds!")
    def check_balance():
        print(f"Current balance: {balance}")
    return deposit, withdraw, check_balance
deposit, withdraw, check = bank_account(100)
deposit(50)
withdraw(30)
check()
```

```python
#Task1 convert
class BankAccount:
    def __init__(self, initial_balance):
        if initial_balance < 0:
            print("Initial balance cannot be negative. Setting balance to 0.")
            self.balance = 0
        else:
            self.balance = initial_balance
    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive!")
        else:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive!")
        elif amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
    def check_balance(self):
        print(f"Current balance: {self.balance}")
account = BankAccount(100)
account.deposit(50)
account.withdraw(30)
account.check_balance()
```

```python
#Task2
class Calculator:
    def add(self, a, b):
        result = a + b
        print(f"{a} + {b} = {result}")
        return result
    def subtract(self, a, b):
        result = a - b
        print(f"{a} - {b} = {result}")
        return result
    def multiply(self, a, b):
        result = a * b
        print(f"{a} * {b} = {result}")
        return result
    def divide(self, a, b):
        if b == 0:
            print("Error: Cannot divide by zero!")
            return None
        result = a / b
        print(f"{a} / {b} = {result}")
        return result
print("Welcome to the calculator!")
print("Choose an operation:")
print("1: Add")
print("2: Subtract")
print("3: Multiply")
print("4: Divide")
calc = Calculator()
choice = input("Enter choice (1/2/3/4): ")
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
if choice == '1':
    calc.add(num1, num2)
elif choice == '2':
    calc.subtract(num1, num2)
elif choice == '3':
    calc.multiply(num1, num2)
elif choice == '4':
    calc.divide(num1, num2)
else:
    print("Invalid choice!")
```

```python
#Task3
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        """Each animal must implement its own sound"""
        pass
    def describe(self):
        print("This is an animal.")
class Dog(Animal):
    def make_sound(self):
        return "Woof!"
class Cat(Animal):
    def make_sound(self):
        return "Meow!"
class Cow(Animal):
    def make_sound(self):
        return "Moo!"
if __name__ == "__main__":
    dog = Dog()
    cat = Cat()
    cow = Cow()
    animals = [dog, cat, cow]
    for animal in animals:
        animal.describe()
        print("Sound:", animal.make_sound())
        print("-" * 20)
```

```python
#Task4
class TextFileReader:
    def __init__(self, file_path):
        """
        Constructor: initializes the file path
        """
        self.file_path = file_path
        self.content = ""

    def read_file(self):
        """
        Opens and reads the file content
        """
        try:
            with open(self.file_path, 'r', encoding='utf-8') as file:
                self.content = file.read()
            print("File read successfully.")
        except FileNotFoundError:
            print("Error: File not found.")
            self.content = ""
        except Exception as e:
            print(f"An error occurred: {e}")
            self.content = ""

    def count_lines(self):
        """
        Returns number of lines in the file
        """
        if not self.content:
            return 0
        return len(self.content.splitlines())

    def count_words(self):
        """
        Returns number of words in the file
        """
        if not self.content:
            return 0
        return len(self.content.split())

    def count_characters(self):
        """
        Returns number of characters in the file
        """
        if not self.content:
            return 0
        return len(self.content)

    def display_content(self):
        """
        Prints the file content
        """
        if not self.content:
            print("No content to display. Make sure to read the file first.")
        else:
            print("File Content:\n")
            print(self.content)
if __name__ == "__main__":
    file_reader = TextFileReader("sample.txt")
    file_reader.read_file()
    file_reader.display_content()
    print("\nStatistics:")
    print("Lines:", file_reader.count_lines())
    print("Words:", file_reader.count_words())
    print("Characters:", file_reader.count_characters())
```

```python
#Task6
def read_txt_file(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            return file.read()
    except FileNotFoundError:
        return "Error: File not found."
    except IOError:
        return "Error: Unable to read file."
    except Exception as e:
        return f"Unexpected error: {e}"
class UserExtractor:
    def __init__(self, file_path):
        self.file_path = file_path
        self.usernames = {}
    def extract_usernames(self):
        content = read_txt_file(self.file_path)
        if content.startswith("Error") or content.startswith("Unexpected"):
            return content
        lines = content.splitlines()
        for line in lines:
            if ":" in line:
                username, password = line.split(":", 1)
                username = username.strip()
                password = password.strip()
                if username not in self.usernames:
                    self.usernames[username] = password
        return self.usernames
if __name__ == "__main__":
    extractor = UserExtractor("users.txt")
    result = extractor.extract_usernames()
    if isinstance(result, dict):
        print("Extracted Usernames:")
        for user, pwd in result.items():
            print(f"Username: {user}, Password: {pwd}")
    else:
        print(result)
```
